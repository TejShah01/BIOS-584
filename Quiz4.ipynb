## Import Data

<font size='4'>

* Three lists named `study_id_ls`, `measure_date_ls`, and `duration_ls` have been initialized.
* Write a for loop to extract and append the values of `study_id`, `measure_date`, and `duration` to `study_id_ls`, `measure_date_ls`, and `duration_ls`, respectively.

```python
location = "/Users/Tej/Documents/GitHub/BIOS-584/data/Quiz4/"

df1 = pd.read_csv(location + "data_1.txt", sep=",")
df2 = pd.read_csv(location + "data_2.txt", sep=",")
df3 = pd.read_csv(location + "data_3.txt", sep=",")
df4 = pd.read_csv(location + "data_4.txt", sep=",")
df5 = pd.read_csv(location + "data_5.txt", sep=",")
df6 = pd.read_csv(location + "data_6.txt", sep=",")

study_id_ls = []
measure_date_ls = []
duration_ls = []

for df in [df1, df2, df3, df4, df5, df6]:
    for i in range(len(df)):
        study_id_ls.append(df.loc[i, "study_id"])
        measure_date_ls.append(df.loc[i, "measure_date"])
        duration_ls.append(df.loc[i, "duration"])


print(study_id_ls)
print(measure_date_ls)
print(duration_ls)
```

<font size='4'>

* Create a new Pandas Dataframe combining three lists and name it `input_df`.

```python
input_df = pd.DataFrame({
    "study_id": study_id_ls,
    "date": measure_date_ls,
    "duration": duration_ls})
```

<font size='4'>

* Show or print `input_df`

```python
import matplotlib.pyplot as plt

#Made graph by accident, the next code chunk has the table

plt.plot(input_df["date"], input_df["duration"])
plt.show()
```

## Data Manipulation

<font size='4'>

* Reorder the dataframe by `measure_date` (earliest to latest) and then by `duration` (shortest to longest).

```python
input_df = input_df.sort_values(["date", "duration"])
```

<font size='4'>

* Create a new column called `relative_time` by calculating the time difference in days between each of the measure date and the earliest one.

```python
input_df["relative_time"] = input_df["duration"] - input_df["duration"].iloc[0]
```

<font size='4'>

* Reset the index column and remove the old index.

```python
input_df = input_df.reset_index(drop=True)
```
